### build demo

#### use command

```python
%%sh
rm ./demo

export CC=gcc-11 CXX=g++-11
g++ -v -std=c++14 -O3 ./demo.cpp \
    -I../../InspireFace/3rdparty/MNN/include \
    -I../../InspireFace/3rdparty/InspireCV/include \
    -I../../InspireFace/cpp/inspireface/include \
    -I../../InspireFace/cpp/inspireface/c_api \
    -I/usr/include/opencv4 \
    -L../../InspireFace/build/inspireface-linux/InspireFace/lib \
    -lInspireFace \
    -lopencv_core -lopencv_imgproc -lopencv_highgui -lopencv_imgcodecs \
    -o ./demo
```

#### use CMakeLists.txt (recommanded)

```python
%%sh
ln -sf CMakeLists_demo.txt CMakeLists.txt
rm demo
mkdir -p ./build
cd ./build
rm -rf *
cmake ..
make
cp demo ../demo
```

```python
%%sh
export LD_LIBRARY_PATH=/kaggle/temp/InspireFace/build/inspireface-linux/InspireFace/lib:$LD_LIBRARY_PATH
./demo
```
