## 3 onnx -> engine

### 3.1 trtexec (failed)

- in case
    ```log
    [05/12/2026-21:50:00] [V] [TRT] Trying to load shared library libnvinfer_builder_resource.so.10.8.0
    [05/12/2026-21:50:00] [E] [TRT] createInferBuilder: Error Code 6: API Usage Error (Unable to load library: libnvinfer_builder_resource.so.10.8.0: libnvinfer_builder_resource.so.10.8.0: cannot enable executable stack as shared object requires: Invalid argument)
    ```
- issue: https://github.com/NVIDIA/TensorRT/issues/2218#issuecomment-1329813418
- add: `LD_PRELOAD=/opt/tensorrt/lib/libnvinfer_builder_resource.so.10.8.0`

````python
! LD_LIBRARY_PATH=/opt/tensorrt/lib:$LD_LIBRARY_PATH \
    ldd /opt/tensorrt/bin/trtexec 
````

````python
! LD_LIBRARY_PATH=/opt/tensorrt/lib:$LD_LIBRARY_PATH \
    LD_PRELOAD=/opt/tensorrt/lib/libnvinfer_builder_resource.so.10.8.0 \
    /opt/tensorrt/bin/trtexec \
    --onnx=scrfd_2.5g_bnkps.onnx \
    --saveEngine=_00_scrfd_2_5g_bnkps_shape160x160_fp16.engine \
    --fp16 \
    --inputIOFormats=fp32:chw \
    --outputIOFormats=fp32:chw \
    --minShapes=input.1:1x3x160x160 \
    --optShapes=input.1:1x3x160x160 \
    --maxShapes=input.1:1x3x160x160 \
    --verbose
````
